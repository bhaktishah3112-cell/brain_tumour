```python
def print_spiral_order(matrix):
    if not matrix.size==0:
        return None

    top = 0
    bottom = len(matrix) - 1
    left = 0
    right = len(matrix[0]) - 1

    while top <= bottom and left <= right:
        # Top row
        for i in range(left, right + 1):
            print(matrix[top][i], end=" ")
        top += 1

        # Right column
        for i in range(top, bottom + 1):
            print(matrix[i][right], end=" ")
        right -= 1

        # Bottom row
        if top <= bottom:
            for i in range(right, left - 1, -1):
                print(matrix[bottom][i], end=" ")
            bottom -= 1

        # Left column
        if left <= right:
            for i in range(bottom, top - 1, -1):
                print(matrix[i][left], end=" ")
            left += 1
import numpy as np
m = np.random.randint(low=5, high=11)
n = np.random.randint(low=5, high=11)
arr = np.random.randint(low=0, high=20, size=(m, n))
for i in range(m):
    for j in range(n):
        print(f"{arr[i][j]:<5}", end="")
    print("")

# You can treat the numpy array as a list in python (for now)
print_spiral_order(arr)
```

```python
def count_word_frequency(sentence):
    freq = {}

    if not sentence:
        return freq

    words = sentence.split()

    for word in words:
        # normalize case and remove punctuation at ends
        word = word.lower().strip(".,!?;:")

        freq[word] = freq.get(word, 0) + 1

    return freq
testcases = [
    "Hello world Hello",
    "Python is fun and Python is easy",
    "One fish two fish red fish blue fish",
    "This is a test. This test is simple.",
    "Repeat repeat Repeat",
    "",
    "Spaces       between      words"
]

for i, test in enumerate(testcases):
    print(f"Test Case {i + 1}:")
    print(f"Input: {test}")
    print("Output:", count_word_frequency(test))
    print("-" * 30)
```

```python
def is_palindrome_iterative(s):
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1

    return True


def is_palindrome_recursive(s):
    if len(s) <= 1:
        return True

    if s[0] != s[-1]:
        return False

    return is_palindrome_recursive(s[1:-1])


test_cases = [
    ("radar", True),
    ("boommoob", True),
    ("world", False),
    ("a", True),
    ("", True),
    ("RadAr", False),
]

for i, (input_str, expected) in enumerate(test_cases):
    print(f"Test Case {i + 1}: Input = '{input_str}'")
    result_iterative = is_palindrome_iterative(input_str)
    result_recursive = is_palindrome_recursive(input_str)
    print(f"  Iterative Output: {result_iterative}")
    print(f"  Recursive Output: {result_recursive}")
    print(f"  Expected Output: {expected}")
    print(f"  Iterative Pass: {result_iterative == expected}")
    print(f"  Recursive Pass: {result_recursive == expected}")
    print("-" * 30)
```

```python
def is_prime(n):
    if n <= 1:
        return False

    if n == 2:
        return True

    if n % 2 == 0:
        return False

    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2

    return True


test_cases = [
    (2, True),
    (3, True),
    (4, False),
    (17, True),
    (19, True),
    (20, False),
    (1, False),
    (0, False),
    (-5, False),
    (97, True)
]

for i, (input_num, expected) in enumerate(test_cases):
    print(f"Test Case {i + 1}: Input = {input_num}")
    result = is_prime(input_num)
    print(f"  Output: {result}")
    print(f"  Expected Output: {expected}")
    print(f"  Pass: {result == expected}")
    print("-" * 30)
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(0, 1, 1000)

mean_val = np.mean(data)
percentile_25 = np.percentile(data, 25)
percentile_75 = np.percentile(data, 75)

plt.figure(figsize=(8,5))
plt.hist(data, bins=20, color='lightpink', edgecolor='black')

plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.2f}')
plt.axvline(percentile_25, color='blue', linestyle='--', linewidth=2, label=f'25th percentile: {percentile_25:.2f}')
plt.axvline(percentile_75, color='green', linestyle='--', linewidth=2, label=f'75th percentile: {percentile_75:.2f}')

plt.text(mean_val, plt.ylim()[1]*0.9, 'Mean', color='red', ha='center')
plt.text(percentile_25, plt.ylim()[1]*0.8, '25th percentile', color='blue', ha='center')
plt.text(percentile_75, plt.ylim()[1]*0.8, '75th percentile', color='green', ha='center')
plt.title('Histogram of Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
